Train

```python
import numpy as np
import pandas as pd

FEATURES = [
    "Setting_1",
    "Setting_2",
    "Sensor_2",
    "Sensor_3",
    "Sensor_4",
    "Sensor_6",
    "Sensor_7",
    "Sensor_8",
    "Sensor_9",
    "Sensor_11",
    "Sensor_12",
    "Sensor_13",
    "Sensor_14",
    "Sensor_15",
    "Sensor_17",
    "Sensor_20",
    "Sensor_21"
]


df = pd.read_csv("processed_train.csv")

rng = np.random.default_rng(42)


random_scores = rng.random(
    size=(len(df), len(FEATURES))
)

missing_rates = [0.05, 0.10, 0.20, 0.30]

missing_datasets = {}

for rate in missing_rates:

    df_missing = df.copy()


    mask = random_scores < rate

    feature_values = df_missing[FEATURES].to_numpy().copy()

    feature_values[mask] = np.nan

    df_missing[FEATURES] = feature_values

    missing_datasets[rate] = df_missing

    print(
        f"{int(rate*100)}% dataset:",
        df_missing[FEATURES].isna().sum().sum(),
        "missing values"
    )
```

```python
for rate, data in missing_datasets.items():

    pct = int(rate * 100)

    data.to_csv(
        f"processed_train_missing_{pct}.csv",
        index=False
    )
```

```python
for rate, data in missing_datasets.items():

    missing_pct = (
        data[FEATURES].isna().sum().sum()
        / data[FEATURES].size
        * 100
    )

    print(
        f"Requested: {rate*100:.0f}% | "
        f"Actual: {missing_pct:.2f}%"
    )
```

Test

```python
test_df = pd.read_csv("Dataset/processed_test.csv")

rng_test = np.random.default_rng(123)

test_scores = rng_test.random(
    size=(len(test_df), len(FEATURES))
)

for rate in missing_rates:

    test_missing = test_df.copy()

    mask = test_scores < rate

    values = test_missing[FEATURES].to_numpy().copy()
    values[mask] = np.nan

    test_missing[FEATURES] = values

    test_missing.to_csv(
        f"processed_test_missing_{int(rate*100)}.csv",
        index=False
    )
```
